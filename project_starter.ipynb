## 2. Check Gurobi

As in the assignments, maximize a variable bounded above by 1. The expected objective is `1.0`. A license error here needs to be resolved before building the project model. This small check does not confirm that the license supports a full-size project model.


```python
with gp.Model("setup_check") as check:
    check.setParam("OutputFlag", 0)
    x = check.addVar(lb=0, ub=1.0, name="x")
    check.setObjective(x, GRB.MAXIMIZE)
    check.optimize()

    if check.Status != GRB.OPTIMAL:
        raise RuntimeError(f"Setup check did not solve optimally. Status: {check.Status}")
    print("Test objective:", check.ObjVal)
    assert abs(check.ObjVal - 1.0) < 1e-6
    print("Gurobi setup check passed.")
```
